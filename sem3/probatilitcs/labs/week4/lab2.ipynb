```python
n = 100
counts = {total : 0 for total in range(3,19)}

for _ in range(n):
    roll = sum(randint(1,6) for _ in range(3))
    counts[roll] += 1

plot(list(counts.keys()),list(counts.values()),'xb')
xticks(list(counts.keys()))
yticks(list(counts.values()))
grid()
show()
```
