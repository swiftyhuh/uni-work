```python
from random import sample

def prob_cond(nr_sim):
    countA, countB = 0, 0
    for _ in range(nr_sim):
        drawn_balls = sample(['red','blue','green'],counts=[5,3,2],k=3)
        if 'red' in drawn_balls:
            countA+=1
            if len(set(drawn_balls))==1:
                countB+=1
    return countB/countA
prob_cond(5000)
```

```python
# Bernoulli X(0 1   E Bern(p)
#             1-p p) where p - prob of succes

# Binomial X = number of succeses in n Bernoulli trials
#          X belongs to Bino(n, p) - n numbers of trials, p - prob of succes in a single trial

# X(0 1 2 ... k ... n
#   C(k,n) * p(k, II) * (1-p)^p-k
#   P(,II)(x=k)
#   binom.pmf(k,n,p))

# P(x <= k) = Binom.cdf(k, n, p)

# problem 3
# X belongs to (5, 6/10)
# binom.rvs(n, p, size = ...)


from scipy.stats import binom
from matplotlib.pyplot import hist, grid, show, legend

# help('scipy.stats.binom')

x = binom.rvs(5, 6/10, size = 1000)
print(x)
bin_edges = [k+0.5 for k in range(-1,6)]
hist(x, bin_edges, density = True, rwidth = 0.9, color = 'green', edgecolor = 'black',
    alpha = 0.5, label = 'relative frequencies')

distribution = dict([(i, binom.pmf(i, 5, 6/10)) for i in range(0,6)])
bar(distribution.keys(), distribution.values(), width = 0.85, color = 'red', edgecolor = 'black', alpha = 0.6, label = 'probabilities')

legend(loc = 'lower left')
grid()
show()


```

```python
# total prd formula
# (Bk)k = events that forma partition (they can't happen at the same time)
# prob event B1 or B2 or last B should be exactly 1
# we have another event A P(A) = sum(k=1 to n) of all the case of P(Bk) * P(A|Bk)
# Bk = "getting k in step 1 (first line)"
# P(Bk) = 1 / 100
# P(A) = "the algo outputing 100"
# to compute P(A) we need to compute P(A|Bk)
# P(A|Bk) = "what is the prob that choosing one element from Bk to be 100"
# prob that choosing the number 100 out of the list, without replacement
# the first time in the list and choosing one number and i want to check if it is100
# P(A|Bk) = Comb (99, k-1) / Comb(100, K)

from math import comb

total = 0
for k in range(1, 101):
    p_Bk = 1/100
    p_A_given_Bk = comb(99, k-1) / comb(100, k)
    total += p_Bk * p_A_given_Bk

print(total)
```
