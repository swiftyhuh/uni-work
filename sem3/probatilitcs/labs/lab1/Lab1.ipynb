```python
from math import perm, comb
from itertools import permutations, combinations
def get_arrangements(w, k,total_number=False, at_random=False):
    if total_number == False and at_random == False:
        return list(permutations(w, k))
    elif total_number == True and at_random == False:
        return len(list(permutations(w, k)))
    elif total_number == False and at_random == True:
        return sample(list(permutations(w, k)), 1)

def get_combinations(w, k,total_number=False, at_random=False):
    if total_number == False and at_random == False:
        return list(combinations(w, k))
    elif total_number == True and at_random == False:
        return len(list(combinations(w, k)))
    elif total_number == False and at_random == True:
        return sample(list(combinations(w, k)), 1)




# get_arrangements("word",2)
# get_arrangements("word",2,total_number=True)
# get_arrangements("word",2,at_random=True)
# get_combinations("word",2)
# get_combinations("word",2,total_number=True)
# get_combinations("word",2,at_random=True)
```
